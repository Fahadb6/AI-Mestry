```python
import pandas as pd
import numpy as np
from datasets import load_dataset
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
import matplotlib.pyplot as plt
```

```python
df['job_posted_month'] = df['job_posted_date'].dt.month
```

```python
monthly_counts = df.job_posted_month.value_counts()
monthly_counts = monthly_counts.sort_index()
monthly_counts
```

```python
date_counts = df.job_posted_date.value_counts().sort_index()
```

```python
plt.plot(date_counts.index , date_counts)
```

```python
plt.plot(monthly_counts.index , monthly_counts)
```

```python
df.job_title_short.value_counts()
```

```python
jobs_counts = df.job_title_short.value_counts()#.head(4)
# I can use Head() to put how many values I want to count like value_counts().head(3) --> it will give me 3 
```

```python
plt.bar(jobs_counts.index , jobs_counts)
```

```python
jobs_counts = jobs_counts.sort_values(ascending=False) #
plt.barh(jobs_counts.index , jobs_counts)
plt.title('Posting By Job Title')
# I delete the Head above to see all Jobs here 
plt.show()
```

```python
plt.bar(jobs_counts.index , jobs_counts)
plt.title('Posting By Job Title')
plt.ylabel('Count of Job Posting')
plt.xticks(rotation = 45 , ha='right');
```
